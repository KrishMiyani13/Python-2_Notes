```python
import pandas as pd
import numpy as np
```

```python
data = {'ID':[1,2,3,4],'Nmae':['Aumit','Priya','Raj','Rahul'],'Marks':[80,75,80,85],'City':['Ahmedabad','Rajkot','Bhavnagr','Surat']}
df = pd.DataFrame(data)
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df = df.bfill()
df
```

```python
df.fillna(df.mean(numeric_only=True))
```

```python
df.fillna(df.median(numeric_only=True))
```

```python
df.fillna(df.mode(numeric_only=True))
```

```python
print(df)
df = df.fillna(df.mode().iloc[0])

df
```
