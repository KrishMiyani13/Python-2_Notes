```python
df.isna().sum()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
df.isna().sum()
```

```python
df['fare'] = df['fare'].fillna(df['fare'].median())
df.isna().sum()
```

```python
df['embarked'].unique()
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df.isna().sum()
```

```python
df['cabin'].unique()
```

```python
df['cabin'] = df['cabin'].fillna(method = 'bfill')
df.isna().sum()
```

```python
df['cabin'] = df['cabin'].fillna(method = 'ffill')
df.isna().sum()
```
