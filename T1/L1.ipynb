```python
import pandas as pd
import numpy as np
data = [1,2,3,4,5]
S = pd.Series(data)
print(S)
```

```python
data2 = (1,2,3,4,5,6)
s = pd.Series(data2)
print(s)
```

```python
data3 = {'A':1,'B':2}
s = pd.Series(data3)
print(s)
```

```python
data = [1,2,3,4,5]
S = pd.Series(data,index=['A','B','C','D','E'])
print(S)
```

```python
data = [1,2,3,4,5]
S = pd.Series(data,index=['A','B','C','D','E'],name="Numbers",dtype='float')
print(S)
```

```python
s = pd.Series(5,index=[0,1,2,3,4],copy=True)
print(s)
```

```python
data = [np.nan,70,85,77,87]
S = pd.Series(data,index=['A','B','C','D','E'],name='Number',copy=True,dtype='float')
print(S)
print('*'*30)
print(S.values)
print('*'*30)
print(S.index)
print('*'*30)
print(S.shape)
print('*'*30)
print(S.size)
print('*'*30)
print(S.ndim)
print('*'*30)
print(S.hasnans)
print('*'*30)
print(S.name)
print('*'*30)
print(S.dtypes)
print('*'*30)
print(S.is_unique)
print('*'*30)
```

```python
s = pd.Series(data,index=['A','B','C','D','E'])
print(s.loc['A'])
print(s.loc[])
```

```python
s = pd.Series(data)
print(s.loc[0])
print(s.iloc[2])
s.iloc[0] = 999
print(s.iloc[0])
print(s.loc[2:4])
print(s.iloc[3:5])
s.iloc['B':'D']
s.loc['B':'D']
```

```python
print(s.iloc[2:10:2])
print(s.iloc[[2,4]])
```

```python
s1 = s>50
print(s[s1])
```

```python
print(s.loc[s>50])
```
