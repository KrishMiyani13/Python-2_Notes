```python
employes = pd.DataFrame({
    'emp_id':[1,2,3,4],
    'name':['A','B','C','D'],
    'dept_id':[10,20,10,30]
})
department = pd.DataFrame({
    'dept_id':[10,20,40],
    'name':['HR','IT','FINACE']
})
pd.merge(employes,department,on='dept_id')
```

```python
pd.merge(employes,department,on='dept_id',how='inner')
```

```python
pd.merge(employes,department,on='dept_id',how='outer')
```

```python
pd.merge(employes,department,on='dept_id',how='left')
```

```python
pd.merge(employes,department,on='dept_id',how='right')
```

```python
pd.merge(employes,department,how='cross')
```

```python
pd.merge(employes,department,left_on='dept_id',right_on='dept_id')
```

```python
pd.merge(employes,department,left_on='dept_id',right_on='dept_id',how='outer')
```

```python
pd.merge(employes,department,left_on='dept_id',right_on='dept_id',how='outer',indicator=True)
```

```python
pd.merge(employes,department,left_on='dept_id',right_on='dept_id',how='outer',indicator=True,sort=True,copy=True,suffixes=['_emp','_dpt'])
```

```python
pd.merge(employes,department,on='dept_id',left_index=True)
```
