```python
url = "https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets"
res = requests.get(url)
soup = BeautifulSoup(res.text,"html.parser")
```

```python
name = soup.find_all('a',class_="title")
```

```python
product_name = [i.text.strip() for i in name]
print(product_name)
```

```python
price = soup.find_all('h4',class_="price float-end card-title pull-right")
```

```python
price_all = [i.text.strip() for i in price]
print(price_all)
```

```python
review = soup.find_all('p',class_="review-count float-end")
review_product = [i.text.strip() for i in review]
print(review_product)
```

```python
description = soup.find_all('p',class_="description card-text")
desc_product = [i.text.strip() for i in description]
print(desc_product)
```

```python
import pandas as pd
```

```python
data = pd.DataFrame({
    "prduct_name":product_name,
    "price":price_all,
    "review":review_product,
    "description":desc_product
})
data
```
