```python
from bs4 import BeautifulSoup
```

```python
html = """
<html>
    <body>
        <h1 class="demo">Hello</h1>
        <p class="par1" id="test">First</p>
        <p class="par2" >Second</p>

        <a href="www.github.com/Krishmiyani13" class="link1">Krishmiyani13</a>
        <a href="https://in.linkedin.com/"class="link2">Linkdin</a>
    </body>
</html>
"""
html
```

```python
soup = BeautifulSoup(html,"html.parser")
soup
```
