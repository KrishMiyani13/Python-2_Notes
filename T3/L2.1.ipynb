```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
base_url = "https://books.toscrape.com/catalogue/page-{}.html"
title = []
price = []
reating = []
for page in range(1,6):
    url = base_url.format(page)
    res = requests.get(url)
    soup = BeautifulSoup(res.text,"html.parser")
    books = soup.find_all("article",class_="product_pod")
    
    for book in books:
        title.append(book.h3.a['title'])
        price.append(book.find('p',class_="price_color").text)
        reating.append(book.p["class"][1])
print(title)
print(price)
print(reating)  

datafram = pd.DataFrame({"Book_title":title,
                         "Book_price":price,
                         "Book_Reating":reating})
```
