```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
base_url = "https://subslikescript.com/movies?page={}"
title = []
link = []
for page in range(1,6):
    url = base_url.format(page)
    print(url)
    res = requests.get(url)
    print(res)
    soup = BeautifulSoup(res.text,"html.parser")
    movie = soup.find("ul",class_="scripts-list")
    li = movie.find_all("li")
    for i in li:
        title.append(i.a["title"].split("'")[1])
        link.append("https://subslikescript.com"+i.a['href'])
```

```python
dataFream = pd.DataFrame({"Title":title,'Links':link})
dataFream
```
