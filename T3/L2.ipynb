```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url = requests.get("https://www.iplt20.com/auction")
```

```python
url.status_code
```

```python
soup = BeautifulSoup(url.text,"html.parser")
```

```python
year = soup.find("ul",class_='year-auction nav')
year
```

```python
y_2019 = year.find_all('a')
y_2019[7].text
```

```python
table = soup.find("table",class_="ih-td-tab w-100 auction-tbl")
rows = table.find_all("tr")
th = rows[0].find_all("th")
```

```python
columns = [i.text.strip() for i in rows[0].find_all('th')[2:]]
print(columns)
```

```python
data = []
for i in rows[1:]:
    td = i.find_all('td')[1:]
    data.append([j.text.strip() for j in td])
data
```

```python
datafram = pd.DataFrame(data,columns=columns)
datafram
```
