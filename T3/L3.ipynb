```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
webside = "Fact-checks _ PolitiFact.html"
file = open(webside,encoding='utf-8')

res = file.read()
soup = BeautifulSoup(res,"html.parser")
soup
```

```python
Statement_List = soup.find_all('article',class_="m-statement")
Statement = []
User = []
Date = []
for i in Statement_List:
    User.append(i.a.text.strip())
    Statement.append(i.find("div",class_="m-statement__quote").a.text.strip())
    date = (i.find('div',class_="m-statement__desc").text.strip().split(" "))
    Date.append(date[2]+" / "+date[3].split(",")[0]+" / "+date[4])
print(Statement)
print(User)
print(Date)
```

```python
dataFram = pd.DataFrame({"Sources":User,"Statement":Statement,"Date":Date})
dataFram
```

```python
dataFram.sort_values('Rating',ascending=False)
```
