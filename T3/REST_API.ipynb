```python
url1 = "http://localhost:8000/api/players/"


data = {
    "name":"thaluu",
    "runs":12000,
    "innings":120
}

res = requests.post(url1,json = data)
print(res.status_code)
```

```python
url1 = "http://localhost:8000/api/players/8/"


data = {
    "name":"thaluu",
}

res = requests.patch(url1,json = data)
print(res.status_code)
```

```python
params = {'page':2,'limit':3}

url = "http://localhost:8000/api/players/"

res = requests.get(url,params=params)

print(res.json())
```
